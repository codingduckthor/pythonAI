```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data = {
    "ads": [5, 10, 15, 20, 30, 40, 50, 60, 75, 90, 110, 130, 160, 200],
    "sales": [20, 35, 50, 65, 90, 120, 150, 170, 210, 250, 290, 320, 360, 400],
    "subscription_time_days": [60, 55, 50, 48, 45, 40, 35, 33, 30, 28, 25, 23, 20, 18]
}

df = pd.DataFrame(data)

x = df[["ads"]]
y = df[["sales", "subscription_time_days"]]

model = LinearRegression()
model.fit(x, y)

predictions = model.predict(x)

sorted_idx = x["ads"].argsort()

x_sorted = x.iloc[sorted_idx]
pred_sorted = predictions[sorted_idx]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(x, y["sales"], label="Real sales")
axes[0].plot(x_sorted, pred_sorted[:, 0], label="Regression")
axes[0].set_title("Sales vs Ads")
axes[0].set_xlabel("Ads")
axes[0].set_ylabel("Sales")
axes[0].legend()

axes[1].scatter(x, y["subscription_time_days"], label="Real subscription")
axes[1].plot(x_sorted, pred_sorted[:, 1], label="Regression")
axes[1].set_title("Subscription vs Ads")
axes[1].set_xlabel("Ads")
axes[1].set_ylabel("Days")
axes[1].legend()

plt.show()
```
