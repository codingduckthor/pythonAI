```python
# 13. Подключить torch.
import torch
import torch.nn as nn
# 25. Установить transformers
from transformers import pipeline

# 14. Вывести версию PyTorch.
print("pyTorch version:", torch.__version__)

# 15.Проверить GPU.
gpu = torch.cuda.is_available()
print("isgpu avaliable?", gpu)

# 16.Создать device.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# 17.Создать tensor размером 3x4 с dtype=torch.float32
X = torch.tensor([
    [5.1, 3.5, 1.4, 0.2],
    [6.0, 2.9, 4.5, 1.5],
    [7.1, 3.0, 5.9, 2.1]
], dtype=torch.float32)

# 18.Вывести shape, dtype, mean, sum.
print("shape:", X.shape)
print("type:", X.dtype)
print("mean value:", X.mean())
print("total sum:", X.sum())

# 19.Создать модель: Linear(4, 8), ReLU, Linear(8, 3)
model = nn.Sequential(
    nn.Linear(4, 8),
    nn.ReLU(),
    nn.Linear(8, 3),
)

# 20.Создать 3 объекта с 4 признаками.
# в задании 17

# 21. Пропустить данные через модель.
output = model(X)

# 22. Вывести output и output.shape.
print("\nOutput:")
print(output)
print("\nOutput shape:")
print(output.shape)

# 23. Получить predictions через torch.argmax(output, dim=1).
predictions = torch.argmax(output, dim=1)
print("\nPredictions:")
print(predictions)

# 24.Письменно объяснить, почему модель выдает 3 числа на каждый объект.
# модель, на выходе, имеет 3 нейрона. эти нейроны дают 3 оценки на каждый поступающий к ним обьект

# 26. Создать classifier
classifier = pipeline("sentiment-analysis")

# 27. Создать список из 10 отзывов
reviews = [
    "I love this phone, the battery is amazing.",
    "This laptop is very slow and disappointing.",
    "The movie was fantastic and emotional.",
    "I hate this application, it crashes every time.",
    "The lesson was clear and useful.",
    "Terrible customer service experience.",
    "The food was delicious and fresh.",
    "This game is boring and repetitive.",
    "Excellent quality and fast delivery.",
    "The hotel room was dirty and noisy."
]

# 28. Для каждого отзыва вывести text, label, score.
print("\nSentiment Analysis Results:\n")

for review in reviews:
    result = classifier(review)[0]

    print("Text:", review)
    print("Label:", result["label"])
    print("Score:", round(result["score"], 4))
    print("-" * 50)

# 29-30. Создать функцию analyze_review(text)
def analyze_review(text):
    result = classifier(text)[0]

    return {
        "text": text,
        "label": result["label"],
        "score": round(result["score"], 4)
    }

# 31. Проверить функцию на 5 новых текстах.
new_reviews = [
    "This camera works perfectly.",
    "The update ruined the application.",
    "I really enjoyed this book.",
    "The product quality is terrible.",
    "Amazing experience and friendly staff."
]

print("\nFunction Testing:\n")

for review in new_reviews:
    analysis = analyze_review(review)
    print(analysis)
```
