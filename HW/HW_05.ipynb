```python
import numpy as np
import pandas as pd

data = {
    "title": [
        "Inception",
        "The Dark Knight",
        "Interstellar",
        "Parasite",
        "The Matrix",
        "Forrest Gump",
        "Gladiator",
        "Titanic"
    ],
    "release_year": [2010, 2008, 2014, 2019, 1999, 1994, 2000, 1997],
    "genre": [
        "Sci-Fi", "Action", "Sci-Fi", "Thriller",
        "Sci-Fi", "Drama", "Action", "Romance"
    ],
    "rating": [8.8, 9.0, 8.6, 8.5, 8.7, 8.8, 8.5, 7.9],
    "duration_min": [148, 152, 169, 132, 136, 142, 155, 195]
}

movies = pd.DataFrame(data)

print(movies)
```

```python
print(movies.head())
```

```python
print(movies.columns)
```

```python
print(movies["title"], "\n")
print(movies["release_year"], "\n")
print(movies["genre"])
```

```python
print(movies.iloc[0:3], "\n")
print(movies.iloc[4:], "\n")
print(movies.iloc[2, 3])
```

```python
print(movies[movies["rating"] < 8], "\n")
print(movies[movies["duration_min"] > 140], "\n")
print(movies[(movies['genre'] == "Action") & (movies['release_year'] > 2007)])
```

```python
print(movies.sort_values("rating"), "\n")
print(movies.sort_values("duration_min", ascending=False))
```

```python
movies["seen"] = np.random.choice([True, False], size=len(movies))

print(movies)
```

```python
print(movies["rating"].mean().round(2), "\n")
print(movies["duration_min"].mean())
```

```python
print(movies.groupby("genre")["title"].count())
```
