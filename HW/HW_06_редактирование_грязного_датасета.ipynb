```python
import pandas as pd
import numpy as np
# датасет с урока
data = {
    "student_id": [101,102,103,104,105,106,107,108],
    "name": ["Ali", "Aruzhan", "Mansur", "Dana", "Dana", None, "Nurlan", "Aizhan"],
    "age":[17,18,np.nan,17,17,19,"18",20],
    "city":["aktobe", "Astana", "Aktobe",None,None,"Aktobe","Almaty", "astana"],
    "math_score": [85,90,78,None,None,92,88,"91"],
    "attendance": [95,99,70,100,100,None,85,93],
    "passed": ["yes","yes","no","yes","yes","yes","no","yes"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# тут выбрал колонки с числами и заменил None на NaN чтобы модели было легче работать
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["math_score"] = pd.to_numeric(df["math_score"], errors="coerce")
df["attendance"] = pd.to_numeric(df["attendance"], errors="coerce")

# тут все строчные колонки приведены к нижнему регистру и удалены пробелы
df["city"] = df["city"].str.strip().str.lower()
df["name"] = df["name"].str.strip()

print(df)
```

```python
# привидение числовых колонок к усредненному значению
df["age"] = df["age"].fillna(df["age"].median())
df["math_score"] = df["math_score"].fillna(df["math_score"].median())
df["attendance"] = df["attendance"].fillna(df["attendance"].median())

# тоже самое со строками. заполнились "актобе" так как его больше всего. также, если имя не указано замена на "unknown"
df["city"] = df["city"].fillna(df["city"].mode()[0])
df["name"] = df["name"].fillna("Unknown")

# усли есть 2 одинаковые строки - удаляются
df = df.drop_duplicates()

print(df)
print("======================")
print(df.info())
print("======================")
print(df.isnull().sum())
```
